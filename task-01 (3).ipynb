Создайте класс Point. У этого класса должны быть

    - метод set_coordinates, который принимает координаты по x и по y, и сохраняет их в экземпляр класса соответственно в атрибуты x и y 
    - метод get_distance, который обязательно принимает экземпляр класса Point и возвращает расстояние между двумя точками. У метода get_distance есть необязательный аргумент - dist, который задает расстояние:
    евклидово(по умолчанию), Чебышева, городских кварталов.
В случае, если в данный метод передается не экземпляр класса Point необходимо вывести сообщение "Передана не точка" и поднять исключение ArgumentError.


```python
from math import hypot


class ArgumentError(TypeError):
    """Ошибка типа аргумента, указанная в условии задания."""


class Point:
    def set_coordinates(self, x, y):
        self.x = x
        self.y = y

    def get_distance(self, other, dist="euclidean"):
        if not isinstance(other, Point):
            print("Передана не точка")
            raise ArgumentError("Передана не точка")

        dx = abs(self.x - other.x)
        dy = abs(self.y - other.y)
        if dist in ("euclidean", "евклидово"):
            return hypot(dx, dy)
        if dist in ("chebyshev", "Чебышева", "чебышева"):
            return max(dx, dy)
        if dist in ("manhattan", "городских кварталов"):
            return dx + dy
        raise ValueError("Неизвестный вид расстояния")
```

**Примечание к решению.** В Python нет встроенного `ArgumentError`, поэтому он определен как наследник `TypeError`. Параметр `dist` принимает `"euclidean"` / `"евклидово"`, `"chebyshev"` / `"Чебышева"` и `"manhattan"` / `"городских кварталов"`. Координаты задаются методом `set_coordinates` перед вычислением расстояния.
